```python
L = [1.0, 2.5, 0.8, 5.0, 10.0]
d = [10, 25, 15, 40, 60]
```

```python
# Para tranformar las listas en arreglos se usa la función np.array
array_L = np.array(L) # arreglo de L
array_d= np.array(d) # arreglo de d
```

```python
array_L_W = array_L*3.828*(10**26) # transformación de luminosidad solar a vatios
array_d_m = array_d*3.0857*(10**16) # transformación de parsec a metros
# La función flujo utiliza los arreglos de L y d, devolviendo otro arreglo en [Wm**-2]
def flujo(L, d):
    F = L/(4*np.pi*(d**2))
    return F
Flujo = flujo(array_L_W, array_d_m)
```

```python
# Los indices de las estrellas con mayor o menor flujo se obtinen con la función np.argmin y np.arg.max
indice_max_F = np.argmax(Flujo) #--> 0
indice_min_F = np.argmin(Flujo) #--> 4
# Para el promedio y la media se usan np.mean y np.median respectivamente
flujo_prom = np.mean(Flujo)
flujo_mediana = np.median(Flujo)
```

```python
# se define la condicion de la máscara y se usa en el arreglo del flujo
mask = Flujo > 10**(-11)
Flujo_mask = Flujo[mask]
```

```python
# Ejercicio 2
# se crea un arreglo (np.linspace) de tiempos de 0 a 10 años con 500 mediciones
arreglo_temporal = np.linspace(0,10,500)
# el ruido gaussiano es de desviación estándar 0.002, y del mismo tamaño que la cantidad de mediciones (500)
ruido_gaussiano = np.random.normal(0.0,0.002,500)
# el flujo normal es de 1 en todos los tiempos (500), por ello se usa np.ones
flujo_normal = np.ones(500)
# se agrega el ruido gaussiano al flujo
flujo_ruidoso = flujo_normal + ruido_gaussiano
# el intervalo de los tiempos es de 4.5 a 5.0 años
transito = (arreglo_temporal >= 4.5) & (arreglo_temporal <= 5.0)
# simulacion del flujo en el intervalo
simulacion = flujo_ruidoso[transito] - 0.02
```
